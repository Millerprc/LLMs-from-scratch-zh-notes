## 2.3 将 token 转换为 token ID

- 从这些 token 中，我们可以构建一个由所有唯一 token 组成的词汇表

````python
all_words = sorted(set(preprocessed))
vocab_size = len(all_words)

print(vocab_size)
````

````python
vocab = {token:integer for integer,token in enumerate(all_words)}
````

- 以下是词汇表中前 50 项：

- 下面我们演示使用小型词汇表对一个短样本文本进行分词的过程：

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch02_compressed/07.webp?123" width="500px">

- 现在将所有内容整合到一个分词器类中

````python
class SimpleTokenizerV1:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {i:s for s,i in vocab.items()}
    
    def encode(self, text):
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
                                
        preprocessed = [
            item.strip() for item in preprocessed if item.strip()
        ]
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids
        
    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        # Replace spaces before the specified punctuations
        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)
        return text
````

- `encode` 函数将文本转换为 token ID
- `decode` 函数将 token ID 还原为文本

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch02_compressed/08.webp?123" width="500px">

- 我们可以使用分词器将文本编码（即分词）为整数
- 这些整数之后可以作为 LLM 的输入嵌入

````python
tokenizer = SimpleTokenizerV1(vocab)

text = """"It's the last he painted, you know," 
           Mrs. Gisburn said with pardonable pride."""
ids = tokenizer.encode(text)
print(ids)
````

- 我们也可以将这些整数解码回文本

````python
tokenizer.decode(ids)
````

## 2.4 添加特殊上下文 token

- 添加一些"特殊" token 来标记未知词和文本结尾是很有用的

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch02_compressed/09.webp?123" width="500px">

- 一些分词器使用特殊 token 来为 LLM 提供额外上下文
- 这些特殊 token 包括：
  - `[BOS]`（beginning of sequence，序列起始）标记文本的开始
  - `[EOS]`（end of sequence，序列结束）标记文本的结束（通常用于拼接多个不相关的文本，例如两篇不同的维基百科文章或两本不同的书等）
  - `[PAD]`（padding，填充）当我们使用大于 1 的 batch size 训练 LLM 时使用（我们可能将多个长度不同的文本放到同一批；用填充 token 将较短的文本补齐到最长文本的长度，使所有文本长度相同）
- `[UNK]` 用于表示词汇表中未包含的词

- 注意 GPT-2 不需要上面提到的任何这些 token，它只使用 `` token 来降低复杂度
- `` 与上面提到的 `[EOS]` 类似
- GPT 也使用 `` 进行填充（由于我们通常在批量输入训练时使用掩码，因此不会关注填充 token，所以这些 token 具体是什么无关紧要）
- GPT-2 不使用 `<UNK>` 来表示词汇表外的词；相反，GPT-2 使用字节对编码（BPE）分词器，将词拆分成子词单元，我们将在后面的小节中讨论

- 上面的代码会报错，因为词 "Hello" 不在词汇表中
- 为了处理这种情况，我们可以向词汇表中添加特殊 token，如 `"<|unk|>"` 来表示未知词
- 既然我们已经在扩展词汇表，再添加一个名为 `""` 的 token，它在 GPT-2 训练中用来标记文本的结束（也用于拼接文本之间，例如当我们的训练数据集由多篇文章、多本书等组成时）

````python
all_tokens = sorted(list(set(preprocessed)))
all_tokens.extend(["<|endoftext|>", "<|unk|>"])

vocab = {token:integer for integer,token in enumerate(all_tokens)}
````

- 我们还需要相应地调整分词器，让它知道何时以及如何使用新的 `<unk>` token

````python
class SimpleTokenizerV2:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = { i:s for s,i in vocab.items()}
    
    def encode(self, text):
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        preprocessed = [item.strip() for item in preprocessed if item.strip()]
        preprocessed = [
            item if item in self.str_to_int 
            else "<|unk|>" for item in preprocessed
        ]

        ids = [self.str_to_int[s] for s in preprocessed]
        return ids
        
    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        # Replace spaces before the specified punctuations
        text = re.sub(r'\s+([,.:;?!"()\'])', r'\1', text)
        return text
````

- 我们使用滑动窗口方法，将位置每次向后移动 1：

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch02_compressed/13.webp?123" width="500px">

- 创建 dataset 和 dataloader，从输入文本数据集中提取数据块

````python
from torch.utils.data import Dataset, DataLoader


class GPTDatasetV1(Dataset):
    def __init__(self, txt, tokenizer, max_length, stride):
        self.input_ids = []
        self.target_ids = []

        # Tokenize the entire text
        token_ids = tokenizer.encode(txt, allowed_special={"<|endoftext|>"})
        assert len(token_ids) > max_length, "Number of tokenized inputs must at least be equal to max_length+1"

        # Use a sliding window to chunk the book into overlapping sequences of max_length
        for i in range(0, len(token_ids) - max_length, stride):
            input_chunk = token_ids[i:i + max_length]
            target_chunk = token_ids[i + 1: i + max_length + 1]
            self.input_ids.append(torch.tensor(input_chunk))
            self.target_ids.append(torch.tensor(target_chunk))

    def __len__(self):
        return len(self.input_ids)

    def __getitem__(self, idx):
        return self.input_ids[idx], self.target_ids[idx]
````

````python
def create_dataloader_v1(txt, batch_size=4, max_length=256, 
                         stride=128, shuffle=True, drop_last=True,
                         num_workers=0):

    # Initialize the tokenizer
    tokenizer = tiktoken.get_encoding("gpt2")

    # Create dataset
    dataset = GPTDatasetV1(txt, tokenizer, max_length, stride)

    # Create dataloader
    dataloader = DataLoader(
        dataset,
        batch_size=batch_size,
        shuffle=shuffle,
        drop_last=drop_last,
        num_workers=num_workers
    )

    return dataloader
````

````python
dataloader = create_dataloader_v1(
    raw_text, batch_size=1, max_length=4, stride=1, shuffle=False
)

data_iter = iter(dataloader)
first_batch = next(data_iter)
print(first_batch)
````

- GPT-2 使用绝对位置嵌入（absolute position embeddings），因此我们再创建一个嵌入层即可：

````python
context_length = max_length
pos_embedding_layer = torch.nn.Embedding(context_length, output_dim)

# uncomment & execute the following line to see how the embedding layer weights look like
# print(pos_embedding_layer.weight)
````

````python
pos_embeddings = pos_embedding_layer(torch.arange(max_length))
print(pos_embeddings.shape)

# uncomment & execute the following line to see how the embeddings look like
# print(pos_embeddings)
````

- 要构造 LLM 中使用的输入嵌入，只需将 token 嵌入和位置嵌入相加：

````python
input_embeddings = token_embeddings + pos_embeddings
print(input_embeddings.shape)

# uncomment & execute the following line to see how the embeddings look like
# print(input_embeddings)
````
